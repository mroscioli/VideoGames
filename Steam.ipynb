SP vs MP

```python
import ast

def extract_unique_elements(series):
    elements = set()
    for x in series.dropna():
        try:
            parsed = ast.literal_eval(x)
            if isinstance(parsed, list):
                for item in parsed:
                    elements.add(item.strip())
        except:
            pass
    return elements

# Estrai elementi unici
genres_set = extract_unique_elements(df_b2["genres"])
categories_set = extract_unique_elements(df_b2["categories"])

# Unione
all_labels = sorted(genres_set.union(categories_set))

# Output
len(all_labels), all_labels
```

```python
mp_labels = [
    "Multi-player",
    "Massively Multiplayer",
    "MMO",
    "Online PvP",
    "Online Co-op",
    "LAN PvP",
    "LAN Co-op",
    "PvP",
    "Shared/Split Screen PvP",
    "Shared/Split Screen Co-op",
    "Co-op",
    "Cross-Platform Multiplayer"
]

def classify_mode(categories):
    if not isinstance(categories, str):
        return "Other"

    has_sp = "Single-player" in categories
    has_mp = any(label in categories for label in mp_labels)

    if has_sp and has_mp:
        return "Hybrid (SP+MP)"
    elif has_mp:
        return "Multiplayer only"
    elif has_sp:
        return "Single-player only"
    else:
        return "Other"

df_b2["game_mode"] = df_b2["categories"].apply(classify_mode)

# Controllo distribuzione
df_b2["game_mode"].value_counts()
```

```python
import pandas as pd


if "is_free" not in df_s2.columns:
    df_s2 = df_s2.copy()
    df_s2["is_free"] = df_s2["price"].eq(0)

# Calcolo quota free-to-play PER ANNO
yearly = (
    df_s2.groupby("release_year")
    .agg(
        total_games=("appid", "count"),
        free_games=("is_free", "sum")
    )
    .reset_index()
    .rename(columns={"release_year": "Year"})
)

yearly["Free-to-play share (%)"] = (
    yearly["free_games"] / yearly["total_games"] * 100
).round(2)

# Teniamo SOLO le due colonne richieste + filtro anni
cover_timeline = (
    yearly[
        (yearly["Year"] >= 2006) & (yearly["Year"] <= 2024)
    ][["Year", "Free-to-play share (%)"]]
    .sort_values("Year")
)

cover_timeline
```
